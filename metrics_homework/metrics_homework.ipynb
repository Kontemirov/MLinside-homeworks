Функция для рассчета взвешенной медианы

```python
def weighted_median(data):
    weights = 1 / np.abs(data)
    idx = np.argsort(data)
    sorted_data = data[idx]
    sorted_weights = weights[idx]
    cumsum = np.cumsum(sorted_weights)
    return sorted_data[np.searchsorted(cumsum, 0.5 * cumsum[-1])]
```

```python
pred_values = state.gamma(shape=10, scale=1, size=10000)
target_values = pred_values + state.normal(loc=0, scale=1, size=10000)
```

Регрессионный ROC-AUC

```python
def regression_roc_auc(pred, target):
    n = len(pred)
    total_pairs = n * (n - 1) // 2
    correct_pairs = 0

    for i in range(n):
        for j in range(i + 1, n):
            if (pred[i] < pred[j]) == (target[i] < target[j]):
                correct_pairs += 1

    return 2 * correct_pairs / (n * (n - 1))
```

7. Регрессионный ROC-AUC

```python
roc_auc = regression_roc_auc(pred_values, target_values)

print(f"ROC-AUC: {roc_auc:.8f}")
```

```python
def fast_regression_roc_auc(pred, target):
    step = 1000
    pred_sample = pred[::step]
    target_sample = target[::step]

    n = len(pred_sample)
    total_pairs = n * (n - 1) // 2
    correct_pairs = 0

    for i in range(n):
        for j in range(i + 1, n):
            if (pred_sample[i] < pred_sample[j]) == (target_sample[i] < target_sample[j]):
                correct_pairs += 1

    return 2 * correct_pairs / (n * (n - 1))
```
